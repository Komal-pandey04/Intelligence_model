```
# ASSIGNMENT 1
# CIFAR-10 IMAGE CLASSIFICATION USING ANN/MLP
# WITH HYPERPARAMETER TUNING

!pip install tensorflow seaborn scikit-learn pandas matplotlib -q


# 1. IMPORT LIBRARIES

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.optimizers import Adam, SGD, RMSprop
from tensorflow.keras.utils import to_categorical

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

import time
import random
import os


# ================================================================
# 2. SET RANDOM SEEDS
# ================================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

print("TensorFlow Version:", tf.__version__)


# ================================================================
# 3. LOAD CIFAR-10 DATASET
# ================================================================

print("\nLoading CIFAR-10 dataset...")

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("\nDataset Loaded Successfully!")

print("Training images shape :", x_train.shape)
print("Training labels shape :", y_train.shape)
print("Testing images shape  :", x_test.shape)
print("Testing labels shape  :", y_test.shape)


# ================================================================
# 4. CIFAR-10 CLASS NAMES
# ================================================================

class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

print("\nCIFAR-10 Classes:")
for i, name in enumerate(class_names):
    print(i, "->", name)


# ================================================================
# 5. UNDERSTAND DATASET
# ================================================================

print("\nNumber of training samples:", len(x_train))
print("Number of testing samples :", len(x_test))
print("Image height              :", x_train.shape[1])
print("Image width               :", x_train.shape[2])
print("Number of channels        :", x_train.shape[3])
print("Number of classes         :", len(class_names))


# ================================================================
# 6. DISPLAY SAMPLE IMAGES
# ================================================================

plt.figure(figsize=(12, 8))

for i in range(12):
    plt.subplot(3, 4, i + 1)

    plt.imshow(x_train[i])

    label = int(y_train[i][0])
    plt.title(class_names[label])

    plt.axis("off")

plt.suptitle("Sample CIFAR-10 Images")
plt.tight_layout()
plt.show()


# ================================================================
# 7. CHECK PIXEL VALUES BEFORE NORMALIZATION
# ================================================================

print("\nPixel value before normalization:")
print("Minimum:", x_train.min())
print("Maximum:", x_train.max())


# ================================================================
# 8. NORMALIZE IMAGE PIXELS
# ================================================================
# Original pixel values: 0 to 255
# After normalization: 0 to 1

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("\nPixel value after normalization:")
print("Minimum:", x_train.min())
print("Maximum:", x_train.max())


# ================================================================
# 9. ONE-HOT ENCODE LABELS
# ================================================================

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("\nOne-hot encoded label example:")
print(y_train[0])

print("New training label shape:", y_train.shape)
print("New testing label shape :", y_test.shape)


# ================================================================
# 10. INPUT SIZE
# ================================================================

IMAGE_HEIGHT = 32
IMAGE_WIDTH = 32
CHANNELS = 3

INPUT_SIZE = IMAGE_HEIGHT * IMAGE_WIDTH * CHANNELS

print("\nInput size after Flatten:")
print("32 x 32 x 3 =", INPUT_SIZE)


# ================================================================
# 11. FUNCTION TO BUILD ANN MODEL
# ================================================================

def build_ann(
    learning_rate=0.001,
    activation="relu",
    optimizer_name="adam",
    dropout_rate=0.0
):

    model = Sequential([

        # Convert 32x32x3 image into 3072 values
        Flatten(input_shape=(32, 32, 3)),

        # Hidden Layer 1
        Dense(512, activation=activation),

        # Hidden Layer 2
        Dense(256, activation=activation),

        # Hidden Layer 3
        Dense(128, activation=activation),

        # Optional Dropout
        Dropout(dropout_rate),

        # Output Layer
        Dense(10, activation="softmax")
    ])

    # Select optimizer
    if optimizer_name.lower() == "adam":
        optimizer = Adam(learning_rate=learning_rate)

    elif optimizer_name.lower() == "sgd":
        optimizer = SGD(learning_rate=learning_rate)

    elif optimizer_name.lower() == "rmsprop":
        optimizer = RMSprop(learning_rate=learning_rate)

    else:
        raise ValueError("Invalid optimizer!")

    # Compile model
    model.compile(
        optimizer=optimizer,
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ================================================================
# 12. CREATE INITIAL ANN MODEL
# ================================================================

model = build_ann(
    learning_rate=0.001,
    activation="relu",
    optimizer_name="adam",
    dropout_rate=0.0
)


# ================================================================
# 13. DISPLAY MODEL SUMMARY
# ================================================================

print("\n================ MODEL SUMMARY ================\n")

model.summary()


# ================================================================
# 14. COUNT TRAINABLE PARAMETERS
# ================================================================

total_params = model.count_params()

print("\nTotal Trainable Parameters:", total_params)

print("\nFirst Dense Layer Parameter Calculation:")
print("(3072 x 512) + 512")
print("=", (3072 * 512) + 512)


# ================================================================
# 15. TRAIN INITIAL MODEL
# ================================================================

EPOCHS = 15
BATCH_SIZE = 64

print("\n================ TRAINING INITIAL MODEL ================\n")

start_time = time.time()

history = model.fit(
    x_train,
    y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.1,
    verbose=1
)

training_time = time.time() - start_time

print("\nTraining Time:", round(training_time, 2), "seconds")


# ================================================================
# 16. PLOT TRAINING AND VALIDATION ACCURACY
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()


# ================================================================
# 17. PLOT TRAINING AND VALIDATION LOSS
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()


# ================================================================
# 18. TEST SET EVALUATION
# ================================================================

print("\n================ TEST EVALUATION ================\n")

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("Test Loss     :", round(test_loss, 4))
print("Test Accuracy :", round(test_accuracy, 4))
print("Test Accuracy :", round(test_accuracy * 100, 2), "%")


# ================================================================
# 19. MAKE PREDICTIONS
# ================================================================

print("\nGenerating predictions...")

y_pred_probability = model.predict(
    x_test,
    verbose=0
)

y_pred = np.argmax(
    y_pred_probability,
    axis=1
)

y_true = np.argmax(
    y_test,
    axis=1
)

print("Predictions generated successfully.")


# ================================================================
# 20. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.show()


# ================================================================
# 21. CLASSIFICATION REPORT
# ================================================================

print("\n================ CLASSIFICATION REPORT ================\n")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names
    )
)


# ================================================================
# 22. SHOW WRONG PREDICTIONS
# ================================================================

wrong_indices = np.where(
    y_pred != y_true
)[0]

print("\nNumber of wrong predictions:", len(wrong_indices))

plt.figure(figsize=(12, 8))

for i, index in enumerate(wrong_indices[:12]):

    plt.subplot(3, 4, i + 1)

    plt.imshow(x_test[index])

    actual = class_names[y_true[index]]
    predicted = class_names[y_pred[index]]

    plt.title(
        f"Actual: {actual}\nPredicted: {predicted}"
    )

    plt.axis("off")

plt.suptitle("Wrong Predictions")
plt.tight_layout()
plt.show()


# ================================================================
# 23. GRID SEARCH
# ================================================================
# We test different combinations of:
# Learning Rate
# Batch Size

learning_rates = [
    0.001,
    0.0005
]

batch_sizes = [
    64,
    128
]

grid_results = []

print("\n================================================")
print("                GRID SEARCH")
print("================================================")


for lr in learning_rates:

    for batch in batch_sizes:

        print("\n--------------------------------------------")
        print("Learning Rate:", lr)
        print("Batch Size   :", batch)
        print("--------------------------------------------")

        grid_model = build_ann(
            learning_rate=lr,
            activation="relu",
            optimizer_name="adam",
            dropout_rate=0.0
        )

        start = time.time()

        grid_history = grid_model.fit(
            x_train,
            y_train,
            epochs=10,
            batch_size=batch,
            validation_split=0.1,
            verbose=0
        )

        elapsed = time.time() - start

        grid_test_loss, grid_test_accuracy = grid_model.evaluate(
            x_test,
            y_test,
            verbose=0
        )

        best_val_accuracy = max(
            grid_history.history["val_accuracy"]
        )

        result = {
            "Learning Rate": lr,
            "Batch Size": batch,
            "Best Validation Accuracy": best_val_accuracy,
            "Test Accuracy": grid_test_accuracy,
            "Test Loss": grid_test_loss,
            "Training Time (sec)": elapsed,
            "Parameters": grid_model.count_params()
        }

        grid_results.append(result)

        print(
            "Best Validation Accuracy:",
            round(best_val_accuracy, 4)
        )

        print(
            "Test Accuracy:",
            round(grid_test_accuracy, 4)
        )

        print(
            "Training Time:",
            round(elapsed, 2),
            "seconds"
        )


# ================================================================
# 24. GRID SEARCH RESULTS
# ================================================================

grid_df = pd.DataFrame(grid_results)

print("\n================ GRID SEARCH RESULTS ================\n")

print(grid_df)


# ================================================================
# 25. BEST GRID SEARCH MODEL
# ================================================================

best_grid_index = grid_df[
    "Best Validation Accuracy"
].idxmax()

best_grid = grid_df.loc[
    best_grid_index
]

print("\n================ BEST GRID CONFIGURATION ================\n")

print("Best Learning Rate:",
      best_grid["Learning Rate"])

print("Best Batch Size:",
      int(best_grid["Batch Size"]))

print("Best Validation Accuracy:",
      round(best_grid["Best Validation Accuracy"], 4))

print("Test Accuracy:",
      round(best_grid["Test Accuracy"], 4))


# ================================================================
# 26. RANDOM SEARCH
# ================================================================

random_configs = [

    {
        "learning_rate": 0.001,
        "batch_size": 64
    },

    {
        "learning_rate": 0.0005,
        "batch_size": 128
    },

    {
        "learning_rate": 0.0001,
        "batch_size": 64
    },

    {
        "learning_rate": 0.002,
        "batch_size": 128
    },

    {
        "learning_rate": 0.0005,
        "batch_size": 64
    }
]


random.shuffle(random_configs)

random_results = []

print("\n================================================")
print("                RANDOM SEARCH")
print("================================================")


# Run first 3 random configurations
for config in random_configs[:3]:

    lr = config["learning_rate"]
    batch = config["batch_size"]

    print("\n--------------------------------------------")
    print("Learning Rate:", lr)
    print("Batch Size   :", batch)
    print("--------------------------------------------")

    random_model = build_ann(
        learning_rate=lr,
        activation="relu",
        optimizer_name="adam",
        dropout_rate=0.0
    )

    start = time.time()

    random_history = random_model.fit(
        x_train,
        y_train,
        epochs=10,
        batch_size=batch,
        validation_split=0.1,
        verbose=0
    )

    elapsed = time.time() - start

    random_test_loss, random_test_accuracy = random_model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    best_val_accuracy = max(
        random_history.history["val_accuracy"]
    )

    random_results.append({

        "Learning Rate": lr,

        "Batch Size": batch,

        "Best Validation Accuracy":
            best_val_accuracy,

        "Test Accuracy":
            random_test_accuracy,

        "Test Loss":
            random_test_loss,

        "Training Time (sec)":
            elapsed,

        "Parameters":
            random_model.count_params()
    })


# ================================================================
# 27. RANDOM SEARCH RESULTS
# ================================================================

random_df = pd.DataFrame(
    random_results
)

print("\n================ RANDOM SEARCH RESULTS ================\n")

print(random_df)


# ================================================================
# 28. ACTIVATION FUNCTION COMPARISON
# ================================================================
# Optional section

activations = [
    "relu",
    "sigmoid",
    "tanh"
]

activation_results = []

print("\n================================================")
print("          ACTIVATION FUNCTION COMPARISON")
print("================================================")


for activation in activations:

    print("\nTesting:", activation)

    activation_model = build_ann(
        learning_rate=0.001,
        activation=activation,
        optimizer_name="adam",
        dropout_rate=0.0
    )

    start = time.time()

    activation_history = activation_model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=64,
        validation_split=0.1,
        verbose=0
    )

    elapsed = time.time() - start

    loss, accuracy = activation_model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    best_val = max(
        activation_history.history["val_accuracy"]
    )

    activation_results.append({

        "Activation": activation,

        "Best Validation Accuracy":
            best_val,

        "Test Accuracy":
            accuracy,

        "Training Time":
            elapsed
    })


activation_df = pd.DataFrame(
    activation_results
)

print("\nActivation Comparison:")

print(activation_df)


# ================================================================
# 29. OPTIMIZER COMPARISON
# ================================================================

optimizers = [
    "adam",
    "sgd",
    "rmsprop"
]

optimizer_results = []

print("\n================================================")
print("             OPTIMIZER COMPARISON")
print("================================================")


for optimizer_name in optimizers:

    print("\nTesting:", optimizer_name)

    optimizer_model = build_ann(
        learning_rate=0.001,
        activation="relu",
        optimizer_name=optimizer_name,
        dropout_rate=0.0
    )

    start = time.time()

    optimizer_history = optimizer_model.fit(
        x_train,
        y_train,
        epochs=5,
        batch_size=64,
        validation_split=0.1,
        verbose=0
    )

    elapsed = time.time() - start

    loss, accuracy = optimizer_model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    best_val = max(
        optimizer_history.history["val_accuracy"]
    )

    optimizer_results.append({

        "Optimizer": optimizer_name,

        "Best Validation Accuracy":
            best_val,

        "Test Accuracy":
            accuracy,

        "Training Time":
            elapsed
    })


optimizer_df = pd.DataFrame(
    optimizer_results
)

print("\nOptimizer Comparison:")

print(optimizer_df)


# ================================================================
# 30. TRAIN FINAL MODEL
# ================================================================
# Select the best configuration obtained from Grid Search.

FINAL_LR = float(
    best_grid["Learning Rate"]
)

FINAL_BATCH_SIZE = int(
    best_grid["Batch Size"]
)

print("\n================================================")
print("              FINAL MODEL")
print("================================================")

print("Final Learning Rate:", FINAL_LR)
print("Final Batch Size:", FINAL_BATCH_SIZE)


final_model = build_ann(

    learning_rate=FINAL_LR,

    activation="relu",

    optimizer_name="adam",

    dropout_rate=0.0
)


# ================================================================
# 31. FINAL MODEL TRAINING
# ================================================================

start_time = time.time()

final_history = final_model.fit(

    x_train,

    y_train,

    epochs=15,

    batch_size=FINAL_BATCH_SIZE,

    validation_split=0.1,

    verbose=1
)

final_training_time = time.time() - start_time


print(
    "\nFinal Training Time:",
    round(final_training_time, 2),
    "seconds"
)


# ================================================================
# 32. FINAL MODEL EVALUATION
# ================================================================

final_loss, final_accuracy = final_model.evaluate(

    x_test,

    y_test,

    verbose=0
)


print("\n================ FINAL RESULTS ================\n")

print(
    "Final Test Loss:",
    round(final_loss, 4)
)

print(
    "Final Test Accuracy:",
    round(final_accuracy, 4)
)

print(
    "Final Test Accuracy:",
    round(final_accuracy * 100, 2),
    "%"
)


# ================================================================
# 33. FINAL TRAINING ACCURACY GRAPH
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    final_history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    final_history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Final Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.show()


# ================================================================
# 34. FINAL TRAINING LOSS GRAPH
# ================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    final_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    final_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Final Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)

plt.show()


# ================================================================
# 35. FINAL PREDICTIONS
# ================================================================

final_probabilities = final_model.predict(
    x_test,
    verbose=0
)

final_predictions = np.argmax(
    final_probabilities,
    axis=1
)

actual_labels = np.argmax(
    y_test,
    axis=1
)


# ================================================================
# 36. FINAL CONFUSION MATRIX
# ================================================================

final_cm = confusion_matrix(

    actual_labels,

    final_predictions
)


plt.figure(figsize=(10, 8))

sns.heatmap(

    final_cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names
)

plt.title(
    "Final Model Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.show()


# ================================================================
# 37. FINAL CLASSIFICATION REPORT
# ================================================================

print(
    "\n================ FINAL CLASSIFICATION REPORT ================\n"
)

print(

    classification_report(

        actual_labels,

        final_predictions,

        target_names=class_names
    )
)


# ================================================================
# 38. FINAL SUMMARY TABLE
# ================================================================

summary = pd.DataFrame({

    "Metric": [

        "Architecture",

        "Hidden Layers",

        "Activation",

        "Optimizer",

        "Learning Rate",

        "Batch Size",

        "Epochs",

        "Total Parameters",

        "Training Time (sec)",

        "Test Loss",

        "Test Accuracy"
    ],

    "Value": [

        "512 -> 256 -> 128",

        3,

        "ReLU",

        "Adam",

        FINAL_LR,

        FINAL_BATCH_SIZE,

        15,

        final_model.count_params(),

        round(final_training_time, 2),

        round(final_loss, 4),

        round(final_accuracy, 4)
    ]
})


print(
    "\n================ FINAL SUMMARY ================\n"
)

print(summary.to_string(index=False))


# ================================================================
# 39. SAVE GRID SEARCH RESULTS
# ================================================================

grid_df.to_csv(
    "grid_search_results.csv",
    index=False
)

print(
    "\nGrid Search results saved as: grid_search_results.csv"
)


# ================================================================
# 40. SAVE RANDOM SEARCH RESULTS
# ================================================================

random_df.to_csv(
    "random_search_results.csv",
    index=False
)

print(
    "Random Search results saved as: random_search_results.csv"
)


# ================================================================
# 41. SAVE ACTIVATION RESULTS
# ================================================================

activation_df.to_csv(
    "activation_comparison.csv",
    index=False
)

print(
    "Activation comparison saved as: activation_comparison.csv"
)


# ================================================================
# 42. SAVE OPTIMIZER RESULTS
# ================================================================

optimizer_df.to_csv(
    "optimizer_comparison.csv",
    index=False
)

print(
    "Optimizer comparison saved as: optimizer_comparison.csv"
)


# ================================================================
# 43. SAVE FINAL MODEL
# ================================================================

final_model.save(
    "cifar10_ann_final.keras"
)

print(
    "Final model saved as: cifar10_ann_final.keras"
)


# ================================================================
# 44. SAVE TRAINING HISTORY
# ================================================================

history_df = pd.DataFrame(
    final_history.history
)

history_df.to_csv(
    "training_history.csv",
    index=False
)

print(
    "Training history saved as: training_history.csv"
)


# ================================================================
# 45. FINAL MESSAGE
# ================================================================

print("\n")
print("============================================================")
print("             ASSIGNMENT 1 COMPLETED")
print("============================================================")
print("Dataset              : CIFAR-10")
print("Model                : ANN / MLP")
print("Hidden Layers        : 3")
print("Hidden Units         : 512, 256, 128")
print("Output Classes       : 10")
print("Activation           : ReLU")
print("Output Activation    : Softmax")
print("Optimizer            : Adam")
print("Hyperparameter Tune  : Grid Search + Random Search")
print("Evaluation           : Accuracy, Loss, Confusion Matrix")
print("Extra Comparison     : Activation + Optimizer")
print("============================================================")
```
